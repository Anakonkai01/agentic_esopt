```python
board = [
    [1, 0, 0, 4],
    [0, 4, 1, 0],
    [0, 1, 4, 0],
    [4, 0, 0, 1],
]
```

```python
def is_valid(board, row, col, value): 
    # check row 
    if value in board[row]:
        return False
    
    # check col 
    for r in board[row]: 
        if board[row][col] == value: 
            return False
    
    
    
    # check 2x2 subgrid 
    box_row = (row // 2)*2 
    col_row = (col // 2)*2 
    
    for r in range(box_row, box_row + 2): 
        for c in range(col_row, col_row + 2): 
            if board[r][c] == value:
                return False
    
    return True 


```
